```python
import matplotlib.pyplot as plt
import scienceplots
import pandas as pd
import numpy as np

plt.style.use('science')

plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial']
plt.rcParams['font.size'] = 10
```

```python
df = pd.read_csv("./runs/detect/train/results.csv")
y = df['epoch']
```

```python
df.head()
```

```python
plt.xlabel('Train Box Loss')
plt.ylabel('Epoch')
plt.plot(df['train/box_loss'], y)
plt.show()
```

```python
plt.xlabel('Train Class Loss')
plt.ylabel('Epoch')
plt.plot(df['train/cls_loss'], y)
plt.show()
```

```python
x = np.array(y)
y_data = np.array(df['metrics/mAP50(B)'])
coeffs = np.polyfit(x, y_data, deg=5)

y_pred = np.polyval(coeffs, x)

fig, ax = plt.subplots(figsize=(3, 2), dpi=300)

plt.ylabel('mAP@50')
plt.xlabel('Epoch')

plt.plot(x, y_data, color='blue', alpha=0.5)
plt.scatter(x, y_pred, color='orange', s=1)

plt.savefig("map50.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
x = np.array(y)
y_data = np.array(df['metrics/mAP50-95(B)'])

coeffs = np.polyfit(x, y_data, deg=5)

y_pred = np.polyval(coeffs, x)

fig, ax = plt.subplots(figsize=(3, 2), dpi=300)

plt.ylabel('mAP@50-95')
plt.xlabel('Epoch')

plt.plot(x, y_data, color='blue', alpha=0.5)
plt.scatter(x, y_pred, color='orange', s=1)

plt.savefig("map5090.png", dpi=300, bbox_inches='tight')

plt.show()
```
